- Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo
- Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo izquierdo
- Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo derecho
- Tomar con cuidado: Saber si dos linkedbinarytree son iguales

1)


```python
def contar_un_hijo(T):
    contador = 0
    for p in T.positions():
        if T.num_children(p) == 1:
            contador += 1
    return contador
    
```

2)

```python
def contar_solo_izquierdo(T):
    contador = 0
    for p in T.positions():
        if T.left(p) is not None and T.right(p) is None:
            contador += 1
    return contador
```

3)

```python
def contar_solo_derecho(T):
    contador = 0
    for p in T.positions():
        if T.right(p) is not None and T.left(p) is None:
            contador += 1
    return contador
```
