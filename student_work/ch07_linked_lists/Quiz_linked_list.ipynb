## 0. Implementación base - `SinglyLinkedList`

Ejecuta esta celda primero, sin modificarla.

```python
class SinglyLinkedList:
    class _Node:
        def __init__(self, element, nxt):
            self._element = element
            self._next = nxt

    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def first(self):
        return self._head._element

    def last(self):
        return self._tail._element

    def insert_first(self, e):
        nuevo_nodo = self._Node(e, self._head)
        self._head = nuevo_nodo
        if self._size == 0:
            self._tail = nuevo_nodo
        self._size = self._size + 1

    def insert_last(self, e):
        nuevo_nodo = self._Node(e, None)
        if self.is_empty():
            self._head = nuevo_nodo
        else:
            self._tail._next = nuevo_nodo
        self._tail = nuevo_nodo
        self._size = self._size + 1
```

## 2. Tu solución

Implementa `resolver(L1, L2)`:

1. Concatena `L2` al final de `L1`
2. Retorna el **elemento** del penúltimo nodo de la lista `L1` ya concatenada (el nodo
   justo antes del último).

```python
def resolver(L1, L2):
    if L2.is_empty():
        pass
    elif L1.is_empty():
        L1._head = L2._head
        L1._tail = L2._tail
        L1._size = L2._size
    else:
        L1._tail._next = L2._head
        L1._tail = L2._tail
        L1._size = L1._size + L2._size

    L2._head = None
    L2._tail = None
    L2._size = 0

    nodo_actual = L1._head
    nodo_anterior = None

    while nodo_actual._next is not None:
        nodo_anterior = nodo_actual
        nodo_actual = nodo_actual._next

    return nodo_anterior._element


resultado = resolver(L1, L2)
print(resultado)
```
