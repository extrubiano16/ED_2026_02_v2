```python
def find_min(T:LinkedBinaryTree):
    "Retorna el contenedor donde se encuentra el minimo"    
    p = T.root()
    while T.left(p) != None:
        p = T.left(p)
    return p
```

```python
def es_ordenado(T:LinkedBinaryTree):
    anterior = None
    for p in T.inorder():
        if anterior != None and p.element() < anterior.element():
            return False
        anterior = p
    return True
```
